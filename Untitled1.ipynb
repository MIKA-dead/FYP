```python
import numpy as np

motion = np.load('KIT-ML/new_joint_vecs/walking_001.npy')
print("数据形状:", motion.shape)  # (帧数, 特征维度)
print("特征维度:", motion.shape[1])

# 查看第一帧数据，了解特征分布
print("第一帧前20个特征值:", motion[0, :20])
print("特征值范围统计:")
print("  最小值:", motion.min())
print("  最大值:", motion.max())
print("  均值:", motion.mean())
print("  标准差:", motion.std())
```

```python
# 查看最后 10 个特征
print("最后10个特征值:", motion[0, -10:])

# 检查这些值是否都是 0 或 1（足部接触标签的特征）
# 如果最后几个特征值都是 0 或 1，那它们就是接触标签
contact_candidates = motion[:, -4:]  # 假设最后4维是接触
print("足部接触候选值范围:", contact_candidates.min(), contact_candidates.max())
print("足部接触候选样本（前5帧）:\n", contact_candidates[:5])
```

```python
import numpy as np

motion = np.load('KIT-ML/new_joint_vecs/walking_001.npy')

# 1. 根节点高度（索引2）
root_height = motion[:, 2]
root_height_mean = np.mean(root_height)
root_height_std = np.std(root_height)

print(f"根节点高度均值: {root_height_mean:.3f}")
print(f"根节点高度标准差: {root_height_std:.3f}")

# 2. 足部接触概率（最后4维的平均值，表示整体接触程度）
foot_contact = motion[:, -4:]  # 最后4维
foot_contact_mean = np.mean(foot_contact)  # 整体平均接触概率

print(f"足部接触平均概率: {foot_contact_mean:.3f}")
print(f"足部接触范围: {foot_contact.min():.3f} ~ {foot_contact.max():.3f}")

# 3. 其他关键统计
print("\n总体统计:")
print(f"  特征均值: {np.mean(motion):.3f}")
print(f"  特征标准差: {np.std(motion):.3f}")
```

```python
import numpy as np
import os
from pathlib import Path

KIT_ML_DIR = Path('KIT-ML')
VECS_DIR = KIT_ML_DIR / 'new_joint_vecs'

# 选取不同动作的样本
samples = ['walking_001.npy', '01496.npy']  # 替换为实际文件名

for file in samples:
    motion = np.load(VECS_DIR / file)
    print(f"\n文件: {file}")
    print(f"  形状: {motion.shape}")
    print(f"  最后10维 (第一帧): {motion[0, -10:]}")
    print(f"  最后4维统计: 均值 {np.mean(motion[:, -4:]):.3f}, 标准差 {np.std(motion[:, -4:]):.3f}")
```
